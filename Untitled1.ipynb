```python
import numpy as np #used to perform calcultions for ml
import matplotlib.pyplot as plt #for graphs
%matplotlib inline
#command for jupyter notebook to display the graph inside the notebook
import keras #deep learning library
from keras.models import Sequential
from keras.layers import Dense, Dropout
from sklearn.metrics import confusion_matrix
import seaborn as sns

np.random.seed(0)
```

#Data


```python
from keras.datasets import mnist #to train and test data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
 #x_train= training images
 #y_train= correct labels for them
```

```python
print(x_train.shape, y_train.shape)
print(x_test.shape, y_test.shape) #to find how many of them do we have
```

Visualize examples


```python
num_classes = 10
f, ax = plt.subplots(1, num_classes, figsize=(20,20)) #1 column, num_classes rows and figure size 20, 20

for i in range(0, num_classes):
  sample = x_train[y_train==i][0] #y_train == i takes all the images whose label is 5 and then 0 at the last means it takes the very 1st image
  ax[i].imshow(sample, cmap='gray')
  ax[i].set_title("Label: {}".format(i), fontsize=16)
```

```python
for i in range(10):
  print(y_train[i])
```

```python
y_train = keras.utils.to_categorical(y_train, num_classes)
y_test = keras.utils.to_categorical(y_test, num_classes)
#did one hot encoding here
```

```python
for i in range(10):
  print(y_train[i])
```

Prepare Data

```python
#Normalize data
x_train = x_train/255.0
x_test = x_test/255.0
```

```python
#Reshape data
x_train[0].shape #since mnist image is 28*28 pixel it gives that, means it has 784 pixel values
```

```python
x_train = x_train.reshape(x_train.shape[0], -1) #-1 tells numpy to calculate the dimensions automatically
x_test = x_test.reshape(x_test.shape[0], -1)
print(x_train.shape) #used to flatten the image, originally the image was 28 by 28 but we compressed it into 784 number vector
```

#Create Model - Fully Connected Neural Network


```python
model = Sequential()

model.add(Dense(units =128, input_shape =(784,), activation ='relu'))
model.add(Dense(units =128, activation ='relu'))
model.add(Dropout(0.25))
model.add(Dense(units=10, activation = 'softmax'))

model.compile(loss ='categorical_crossentropy', optimizer ='adam', metrics=['accuracy'])
model.summary()
```

#Train

```python
batch_size = 512 #tells the model to divide the 600000 train into a batch size of 512
epochs =10 #studying the data 10 times
model.fit(x=x_train, y=y_train, batch_size = batch_size, epochs = epochs)
```

#Evaluate

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print("Test Loss: {}, Test Accuracy: {}".format(test_loss, test_accuracy))
```

```python
y_pred = model.predict(x_test)
y_pred_classes = np.argmax(y_pred, axis =1) #to find the position with the largest values
print(y_pred)
print(y_pred_classes)
```

```python

import tensorflow as tf
import numpy as np
import json
import os
import zipfile

# Use your existing trained model.
# If needed, load it with:
# model = tf.keras.models.load_model("/content/mnist_model.keras")

OUT = "/content/mnist_fpga"
os.makedirs(OUT, exist_ok=True)

dense_layers = [
    layer for layer in model.layers
    if isinstance(layer, tf.keras.layers.Dense)
]

expected_shapes = [(784, 128), (128, 128), (128, 10)]

assert len(dense_layers) == 3, (
    f"Expected 3 Dense layers, found {len(dense_layers)}"
)

metadata = {"quantization_scale": 256, "layers": []}

for i, (layer, expected) in enumerate(
    zip(dense_layers, expected_shapes), start=1
):
    weights, biases = layer.get_weights()

    assert weights.shape == expected, (
        f"{layer.name}: expected {expected}, got {weights.shape}"
    )

    # Q8 fixed-point representation in signed 16-bit words.
    wq = np.rint(weights * 256).astype(np.int64)
    bq = np.rint(biases * 256).astype(np.int64)

    assert np.all((-32768 <= wq) & (wq <= 32767)), \
        f"Layer {i} weight overflow"
    assert np.all((-32768 <= bq) & (bq <= 32767)), \
        f"Layer {i} bias overflow"

    wq = wq.astype(np.int16)
    bq = bq.astype(np.int16)

    # Write two's-complement 16-bit hexadecimal words.
    np.savetxt(
        f"{OUT}/weights{i}.hex",
        wq.flatten(order="C").view(np.uint16),
        fmt="%04x"
    )
    np.savetxt(
        f"{OUT}/biases{i}.hex",
        bq.view(np.uint16),
        fmt="%04x"
    )

    metadata["layers"].append({
        "name": layer.name,
        "input_size": int(weights.shape[0]),
        "output_size": int(weights.shape[1]),
        "activation": layer.activation.__name__,
        "weights": f"weights{i}.hex",
        "biases": f"biases{i}.hex"
    })

    print(
        f"{layer.name}: {weights.shape}, "
        f"activation={layer.activation.__name__}"
    )

with open(f"{OUT}/metadata.json", "w") as f:
    json.dump(metadata, f, indent=2)

print("\nExported files:")
print(sorted(os.listdir(OUT)))
```

```python

import numpy as np
import tensorflow as tf

# Load MNIST test data.
(_, _), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_float = x_test.astype(np.float32).reshape(-1, 784) / 255.0
x_fixed = np.rint(x_float * 256).astype(np.int64)

dense_layers = [
    layer for layer in model.layers
    if isinstance(layer, tf.keras.layers.Dense)
]

# Quantize parameters using 8 fractional bits.
params = []
for layer in dense_layers:
    w, b = layer.get_weights()
    wq = np.rint(w * 256).astype(np.int64)
    bq = np.rint(b * 256).astype(np.int64)
    params.append((wq, bq, layer.activation.__name__))

def round_shift_8(acc):
    # Signed round-to-nearest division by 256.
    return np.where(
        acc >= 0,
        (acc + 128) // 256,
        -((-acc + 128) // 256)
    )

def fixed_predict(x):
    a = x.copy()

    for i, (wq, bq, activation) in enumerate(params):
        # Products are Q16; bias is converted to Q16.
        acc = a @ wq + bq * 256
        a = round_shift_8(acc)

        if activation == "relu":
            a = np.maximum(a, 0)
        elif activation == "softmax":
            # Softmax preserves the index of the maximum.
            pass
        elif activation != "linear":
            raise ValueError(f"Unsupported activation: {activation}")

    return np.argmax(a, axis=1)

# Compare floating-point and fixed-point predictions.
float_predictions = np.argmax(model.predict(x_float, verbose=0), axis=1)
fixed_predictions = fixed_predict(x_fixed)

float_acc = np.mean(float_predictions == y_test) * 100
fixed_acc = np.mean(fixed_predictions == y_test) * 100
agreement = np.mean(float_predictions == fixed_predictions) * 100

print(f"Original model test accuracy: {float_acc:.2f}%")
print(f"Fixed-point test accuracy:    {fixed_acc:.2f}%")
print(f"Prediction agreement:         {agreement:.2f}%")
```
